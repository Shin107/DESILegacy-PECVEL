```python
for comb_idx, (combo_name, combo_labels) in enumerate(CUT_COMBINATIONS.items(), start=2):
    north_dir = f'/user/animesh.sah/FP_CUTS/shape_combined_z_north_test/cumulative_combination_{comb_idx}'
    south_dir = f'/user/animesh.sah/FP_CUTS/shape_combined_z_south_test/cumulative_combination_{comb_idx}'

    os.makedirs(north_dir, exist_ok=True)
    os.makedirs(south_dir, exist_ok=True)

    cumulative_N = np.ones_like(cuts_all_N[0], dtype=bool)
    cumulative_S = np.ones_like(cuts_all_S[0], dtype=bool)

    for step, cut_label in enumerate(combo_labels, start=1):
        idx = CUT_LABELS.index(cut_label)
        cumulative_N &= cuts_all_N[idx]
        cumulative_S &= cuts_all_S[idx]

        fitsio.write(
            f'{north_dir}/north_cumulative_step{step}_{cut_label}.fits',
            shape_north_catalog[cumulative_N],
            clobber=True
        )
        fitsio.write(
            f'{south_dir}/south_cumulative_step{step}_{cut_label}.fits',
            shape_south_catalog[cumulative_S],
            clobber=True
        )

    # Optional: final cumulative file for the full combination
    fitsio.write(
        f'{north_dir}/north_cumulative_{combo_name}.fits',
        shape_north_catalog[cumulative_N],
        clobber=True
    )
    fitsio.write(
        f'{south_dir}/south_cumulative_{combo_name}.fits',
        shape_south_catalog[cumulative_S],
        clobber=True
    )
```
