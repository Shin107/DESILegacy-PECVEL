```python
plt.figure(figsize= (16,12))

colors  = plt.cm.viridis(np.linspace(0, 1, 5))
colors2 = plt.cm.winter(np.linspace(0, 1, 5))
i=0
for file in np.sort(glob.glob('/user/animesh.sah/w_theta_results/treecorr_*_MAG_R_*.npy')):
    th,w,er = np.load(file)
    obj = re.search(r'treecorr_(north|south)_(MAG_[A-Z])_(\d+(?:\.\d+)?)_(\d+(?:\.\d+)?)_', file)
    region = obj.group(1)
    band = obj.group(2).lower()
    mag_r_range = f"{obj.group(3)}-{obj.group(4)}"
    plt.title(f"w(theta) for {band}")
    if region == 'north':
        plt.errorbar(th,w,yerr=er,marker='o',ls= 'None',label=f'N:{mag_r_range}',color=colors[i],alpha=0.9)
    else:
        plt.errorbar(th,w,yerr=er,marker='*',ls='None',label=f'S:{mag_r_range}',color=colors2[i%5],alpha=0.6)
    plt.xscale('log')
    plt.yscale('log')
    i+=1
    

plt.legend()
```
